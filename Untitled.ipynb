```python
import os
from pathlib import Path

base = Path("/home/pulipakv/Dravidian/clean_data")
languages = ["telugu", "kannada", "tamil", "malayalam"]

for lang in languages:
    lang_dir = base / lang
    if not lang_dir.exists():
        print(f"{lang}: directory not found")
        continue
    
    total_words = 0
    total_chars = 0
    total_lines = 0
    files = list(lang_dir.glob("*"))
    
    for f in files:
        if f.is_file():
            text = f.read_text(errors="ignore")
            total_words += len(text.split())
            total_chars += len(text)
            total_lines += text.count("\n")
    
    print(f"\n{'='*40}")
    print(f"Language : {lang.upper()}")
    print(f"Files    : {len(files)}")
    print(f"Words    : {total_words:,}")
    print(f"Chars    : {total_chars:,}")
    print(f"Lines    : {total_lines:,}")
```

```python
from pathlib import Path

def count_stats(filepath):
    words, lines, chars = 0, 0, 0
    with open(filepath, "r", errors="ignore") as f:
        for line in f:
            lines += 1
            chars += len(line)
            words += len(line.split())
    return words, lines, chars

base = Path("/home/pulipakv/Dravidian/clean_data")
languages = ["telugu", "kannada", "tamil", "malayalam"]
splits = {
    "train": "*_train_balanced.txt",
    "val":   "*_val_cleaned_final.txt",
    "test":  "*_test_cleaned_final.txt",
}

print(f"{'Language':<12} {'Split':<8} {'Words':>15} {'Lines':>10} {'Size MB':>10}")
print("-" * 60)

for lang in languages:
    for split, pattern in splits.items():
        files = list((base / lang / split).glob(pattern))
        for f in files:
            size_mb = f.stat().st_size / 1e6
            words, lines, chars = count_stats(f)
            print(f"{lang:<12} {split:<8} {words:>15,} {lines:>10,} {size_mb:>10.2f}")
```
